# Izmir Urban Resilience & Essential Services Atlas
Evidence-first district analysis without a composite ranking.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
matches=list(Path('/kaggle/input').rglob('izmir_district_resilience_capacity.csv'))
assert matches, 'Dataset CSV not mounted'
df=pd.read_csv(matches[0])
print('Rows:',len(df),'Districts:',df.district.nunique())
df.head()
```

## Deterministic quality checks


```python
checks={'30_districts':len(df)==30 and df.district.nunique()==30,'unique_ibbs_code':df.ibbs_code.astype(str).is_unique,'population_positive':(df.population_2025>0).all(),'hospital_total_84':int(df.hospital_records.sum())==84,'market_total_269':int(df.street_market_records.sum())==269,'assembly_total_2382':int(df.assembly_area_records.sum())==2382}
pd.Series(checks,name='PASS')
```

## Capacity context - not a ranking
Charts sort values for readability only.


```python
view=df.sort_values('assembly_areas_per_10k')
ax=view.plot(x='district',y='assembly_areas_per_10k',kind='barh',figsize=(9,8),legend=False,title='Emergency assembly-area records per 10k residents')
ax.set_xlabel('source records per 10,000 residents'); ax.set_ylabel('')
plt.tight_layout(); plt.show()
```

```python
view=df.sort_values('weekday_departures_per_1k')
ax=view.plot(x='district',y='weekday_departures_per_1k',kind='barh',figsize=(9,8),legend=False,title='Representative-weekday departures per 1k residents')
ax.set_xlabel('departures per 1,000 residents'); ax.set_ylabel('')
plt.tight_layout(); plt.show()
```

## Interpretation guardrails
- Zero source records are not proof of real-world absence.
- Transit distances are straight-line proxies, not walking time.
- Per-capita values are arithmetic normalizations, not quality scores.
- No composite resilience or livability ranking is published.
